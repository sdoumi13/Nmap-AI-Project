```python
import json
import pandas as pd
from datasets import Dataset, DatasetDict
from sklearn.model_selection import train_test_split

# Load your nmap dataset
with open('final_finetuning_dataset.json', 'r') as f:
    nmap_data = json.load(f)

# Convert to DataFrame
df = pd.DataFrame(nmap_data)
print(f"Loaded {len(df)} nmap command examples")
print("\nSample data:")
print(df.head(3))

# Split dataset (90% train, 10% test)
train_df, test_df = train_test_split(df, test_size=0.1, random_state=42)

print(f"\nTrain set: {len(train_df)} examples")
print(f"Test set: {len(test_df)} examples")

# Save splits
train_df.to_json('nmap_train.jsonl', orient='records', lines=True)
test_df.to_json('nmap_test.jsonl', orient='records', lines=True)
```

```python
import os
import torch
from datasets import Dataset
from transformers import (
    AutoTokenizer, 
    AutoModelForCausalLM, 
    BitsAndBytesConfig, 
    TrainingArguments
)
from peft import LoraConfig
from trl import SFTTrainer

# --- 1. Configuration ---
LOCAL_MODEL_PATH = r"C:\models\phi3_mini"
OUTPUT_DIR = "./phi3-nmap-results"
TRAIN_JSONL = "nmap_train.jsonl"
TEST_JSONL = "nmap_test.jsonl"

# --- 2. Load Dataset ---
train_ds = Dataset.from_json(TRAIN_JSONL)
eval_ds = Dataset.from_json(TEST_JSONL)

# --- 3. Tokenizer Setup (Local Fix) ---
tokenizer = AutoTokenizer.from_pretrained(
    LOCAL_MODEL_PATH, 
    use_fast=True, 
    trust_remote_code=True,
    local_files_only=True  # Essential: skip DNS calls
)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

SYSTEM_MESSAGE = "Output a precise nmap command for the given task."

def format_prompt(example):
    prompt = f"[INST] <<SYS>>\n{SYSTEM_MESSAGE}\n<</SYS>>\n\n{example['input']} [/INST] {example['output']}"
    return {"text": prompt}

train_ds = train_ds.map(format_prompt)
eval_ds = eval_ds.map(format_prompt)

# --- 4. Quantization (VRAM Safe: 8GB) ---
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True, # Added: Save additional 0.5GB VRAM
)

model = AutoModelForCausalLM.from_pretrained(
    LOCAL_MODEL_PATH,
    quantization_config=bnb_config,
    device_map={"": 0},           # Force to GPU to prevent offload crash
    trust_remote_code=True,
    local_files_only=True
)

# --- 5. LoRA Configuration ---
peft_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
)

# --- 6. Training Arguments (Rename fix for modern versions) ---
training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    per_device_train_batch_size=1,     # Lowered for 8GB safety
    gradient_accumulation_steps=8,     # Increase to maintain batch effectiveness
    learning_rate=2e-4,
    fp16=True,
    gradient_checkpointing=True,       # Essential: swap compute for memory
    optim="paged_adamw_32bit",
    eval_strategy="steps",             # Renamed from evaluation_strategy
    save_steps=200,
    logging_steps=50,
    eval_steps=200,
    report_to="none"
)

# --- 7. SFTTrainer (Simplified keyword handling) ---
trainer = SFTTrainer(
    model=model,
    train_dataset=train_ds,
    eval_dataset=eval_ds,
    peft_config=peft_config,
    #dataset_text_field="text",         # Standardizing on 'text' field
    args=training_args,
    # max_seq_length and tokenizer removed to fit older TRL logic if present
)

# --- 8. Train & Save ---
print("[INFO] Starting training...")
trainer.train()

trainer.model.save_pretrained(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print(f"[DONE] Model saved to {OUTPUT_DIR}")
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from peft import PeftModel

# 1. Config paths
BASE_MODEL_PATH = r"C:\models\phi3_mini"
ADAPTER_PATH = "./phi3-nmap-results"

# --- NEW: Quantization for Inference ---
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)

# 2. Load Base Model in 4-bit
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL_PATH, local_files_only=True)
model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL_PATH,
    quantization_config=bnb_config, # Forces 4-bit to fit in 8GB VRAM
    device_map={"": 0},            # Strictly stay on GPU 0
    trust_remote_code=False,        # Use native stable code
    local_files_only=True
)

# 3. Load the LoRA Adapters
model = PeftModel.from_pretrained(model, ADAPTER_PATH)
model.eval()

```

```python
# --- 4. Define Inference Function ---
def generate_nmap(prompt):
    system_msg = "Output a precise nmap command for the given task."
    full_prompt = f"[INST] <<SYS>>\n{system_msg}\n<</SYS>>\n\n{prompt} [/INST]"
    
    inputs = tokenizer(full_prompt, return_tensors="pt").to("cuda")
    with torch.no_grad():
        outputs = model.generate(
            **inputs, 
            max_new_tokens=64, 
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )
    
    decoded = tokenizer.decode(outputs[0], skip_special_tokens=True)
    command = decoded.split("[/INST]")[-1].strip()
    return command

# --- 5. RUN A TEST ---
print(f"RESULT: {generate_nmap('Perform a quiet TCP SYN scan on 10.0.0.5')}")
```
